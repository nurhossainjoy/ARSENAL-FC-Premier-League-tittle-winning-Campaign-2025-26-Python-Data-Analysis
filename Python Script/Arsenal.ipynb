5. Player Position Analysis

Identify each player's position and analyze how many unique players belong to each position.

```python
player_positions = (
    data[["Player", "Position"]]
    .drop_duplicates()
    .sort_values(by="Player")
)

player_positions
```

6. Goal Contribution Analysis

Create a new column called Goal Contribution:
Goal Contribution = Goals + Assists

```python
data["Goal Contribution"] = data["Goals"] + data["Assists"]
top_5_goal_contributors= data.groupby("Player")["Goal Contribution"].sum().sort_values(ascending=False).head(5)
top_5_goal_contributors
```

7. Most Used Players

Find the Top 10 players who played the highest total number of minutes.

```python
most_used_players=data.groupby("Player")["Minutes"].sum().sort_values(ascending=False).head(10)
most_used_players
```

8. Shooting Performance Analysis
Calculate the total number of:
•	Shots 
•	Shots on Target 
•	Goals 
for each player and identify the best attacking players.


```python
top_attacking_players = ( data.groupby("Player")[["Goals", "Shots_On_Target", "Shots"]] .sum() 
.sort_values(by=["Goals", "Shots_On_Target"], ascending=False) )
top_attacking_players
```

9. Shooting Accuracy
Create a new metric:
Shooting Accuracy = (Shots on Target / Total Shots) × 100
Identify the players with the highest shooting accuracy, using a reasonable minimum number of shots.


```python
data["shooting_accuracy"]=data["Shots_On_Target"]/data["Shots"]*100
shooting_accuracy=data.groupby("Player")["shooting_accuracy"].mean().sort_values(ascending=False).dropna()
shooting_accuracy
```

10. Goals per 90 Minutes
Calculate:
Goals per 90 Minutes = (Goals / Minutes Played) × 90
Identify the most efficient goal scorers.


```python
data["goals_per_90_minutes"]=data["Goals"]/data["Minutes"]*90
goals_per_90_minutes=data.groupby("Player")["goals_per_90_minutes"].mean().sort_values(ascending=False).dropna()
goals_per_90_minutes
```

11. Defensive Performance
Create a new metric:
Defensive Contribution = Tackles Won + Interceptions
Find the Top 10 defensive players based on this metric.


```python
data["Difencive_Contribution"]=data["Tackles_Won"]+data["Interceptions"]
Difencive_Contribution=data.groupby("Player")["Difencive_Contribution"].mean().sort_values(ascending=False).dropna()
Difencive_Contribution
```

12. Discipline Analysis
Identify the players with the highest number of:
